```python
#shows how many dupliates values are removed
print(f"  Removed {df_original_shape - df.shape[0]} duplicates")
```

## **Feature Engg**

```python
#dictionary is created to map rating labels and
#replace rating in dataframe with numeric values from dictionary

rating_map = {
    'G':1,
    'PG':2,
    'PG-13':3,
    'R':4,
    'NC-17':5,
    'TV-Y':6,
    'TV-Y7':7,
    'TV-14':8,
    'TV-MA':9
}

df['ratings'] = df['rating'].map(rating_map).fillna(3)
df
```

```python
#created content age
df['content_age'] = 2021 - df['release_year']
```

```python
#calculate content score
df['content_score'] = ( (df['content_age'] * 0.3) + (df['ratings'] * 0.4) )
```

## **Numpy array operation**

```python
#converted columns to numpy arrays
content_score_array = df['content_score'].values
ratings_array = df['ratings'].values
```

```python
#calculate mean & median
mean = np.mean(content_score_array)
median = np.median(content_score_array)

print("Mean of content score",mean)
print("Median of content score", median)
```

```python
#checked high quality content
high_quality = content_score_array > mean
print("High Quality Content items ",high_quality.sum())
```

```python
#assign each score to quality tier
quality_bins = np.array([0, 2, 4, 6, 8, 10])
df['quality_tier'] = np.digitize(df['content_score'], quality_bins)
df
```

## **Visualization using Matplotlib**

```python
#visualize Line chart
plt.plot(df['release_year'],  df['ratings'])

plt.xlabel('release_year')
plt.ylabel('ratings')

plt.title('Releaser year vs ratings')
plt.show()
```

```python
#visualize bar chart
plt.bar(df['release_year'], df['content_score'])

plt.xlabel('release_year')
plt.ylabel('content_score')

plt.title('Release year vs content score')
plt.show()
```

```python
#visualize scatter chart
plt.scatter(df['ratings'], df['release_year'])

plt.title('ratings vs release year')

plt.xlabel('ratings')
plt.ylabel('release_year')

plt.show()
```

```python
#visualize histogram
plt.hist(df['content_score'])

plt.xlabel('content_score')
plt.ylabel('Frequency')

plt.title('Histogram')
plt.show()
```

```python
# @title
#visualize Pie chart with legend and percentage
plot_pie = df['type'].value_counts()
plt.pie(plot_pie,labels=plot_pie.index,autopct='%1.1f%%')   #autopct - shows values in percentage form
plt.legend(plot_pie.index)

plt.title('Pie Chart')
```
